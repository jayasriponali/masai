```python
import pandas as pd

from sklearn.model_selection import train_test_split

from sklearn.linear_model import LogisticRegression

from sklearn.metrics import accuracy_score

from sklearn.metrics import classification_report
```

```python
df = pd.read_csv("/content/churnguard_data (3).csv")
df
```

```python
df.drop(['customerID'],axis = 1,inplace = True)
```

```python
df
```

```python
df.drop_duplicates(inplace = True)
```

```python
df[['gender','PaymentMethod']] = df[['gender','PaymentMethod']].apply(lambda x : x.str.strip())
```

```python
df[['Churn','PhoneService','PaperlessBilling']] = df[['Churn','PhoneService','PaperlessBilling']].apply(lambda x : x.str.strip().str.title())
```

```python
df['Contract'].unique()
```

```python
contract_map = {
    'Month-to-month':'Month-to-month',
    'Two year':'Two year',
    'month to month': 'Month-to-month',
    'One year': 'One year',
    'month-to-month': 'Month-to-month',
    'Monthly': 'Month-to-month',
    'Two Year': 'Two year',
    '1 year': 'One year',
    '2 year': 'Two year',
    'two year': 'Two year',
    'One Year': 'One year',
    'one year': 'One year'
}

```

```python
df['Contract'] = df['Contract'].map(contract_map).values
df['Contract']
```

```python
df['InternetService'].unique()
```

```python
InternetService_map = {
   'Fiber optic': 'Fiber optic',
    'DSl': 'DSL',
   'DSL': 'DSL',
    'dsl': 'DSL',
    'No': 'No',
   'fiber optic': 'Fiber optic',
   'FiberOptic': 'Fiber optic',
    'Fibre optic':'Fiber optic',
   'Dsl': 'DSL',
    'NO': 'No',
   'nan': 'No',
    'no': 'No'
}
```

```python
df['InternetService'] = df['InternetService'].map(InternetService_map).fillna('No')
df['InternetService']
```

```python
df['TotalCharges'] = pd.to_numeric(df['TotalCharges'],errors = 'coerce')
```

```python
df = df[df['tenure'] > 0]
df
```

```python
df = df[(df['MonthlyCharges'] > 10) & (df['MonthlyCharges'] < 200)]
df
```

```python
df['MonthlyCharges'] = df['MonthlyCharges'].fillna(df['MonthlyCharges'].mean())
```

```python
df['TotalCharges'] =  df['TotalCharges'].fillna(df['TotalCharges'].mean())
```

```python
df['tenure'] = df['tenure'].fillna(round(df['tenure'].median(),2))
```

```python
print(f" shape of the cleaned DataFrame {df.shape}")
```

```python
print(f"missing value counts {df.isna().sum().sum()}")
```

```python
df['Churn'] = df['Churn'].map({'Yes': 1, 'No': 0})
```

```python
df = pd.get_dummies(df,['gender', 'PhoneService', 'InternetService', 'Contract', 'PaperlessBilling', 'PaymentMethod'],drop_first = True)
```

```python
X = df.drop(['Churn'],axis=1)
y = df['Churn']
```

```python
X_train,X_test,y_train,y_test = train_test_split(X,y,train_size = 0.80,random_state = 42)
```

```python
model = LogisticRegression(max_iter = 1000)
```

```python
model.fit(X_train,y_train)
```

```python
y_pred = model.predict(X_test)
```

```python
print(f" accuracy score {accuracy_score(y_test,y_pred): .2f}")
```

```python
target_names = ['Stay', 'Churn']
print(classification_report(y_test,y_pred,target_names = target_names))
```
